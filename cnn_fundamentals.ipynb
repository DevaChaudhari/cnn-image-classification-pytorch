### Normalization (min-max scaling)

As you may recall, feature normalization may greatly improve the performance of gradient descent. The <code>Normalize</code> function from Torchvision is the tensor version of the <code>StandardScaler</code> we used before from the *scikit-learn* library (see Notebook 1).

```python
# Check our lobster image details
lobster_tensor.shape, lobster_tensor.min(), lobster_tensor.max()
```

Apply normalization to our tensor with [0,1] range. 

```python
# normalization applies to each channel separately, therefore you need mean and std for each of them
# if you use mean=0.5 and std = 0.5 you will re-scale feature values to [-1,1] interval (improves training of networks)
normalizer = Normalize(mean=[0.5,0.5,0.5], std=[0.5,0.5,0.5])

# tensor with symmetrical range
lobster_tensor_sym = normalizer(lobster_tensor)
```

<code>Normalize</code> shifts the image values by *mean* (as specified in the constructor call; it has nothing to do with the actual mean of the image pixel values) and divides by *std* as you define in the constructor call (again this number has no connection to the actual standard deviation of the image pixel values). We may write it as a formula:
$$\large\text{image} = \large\frac{\text{image}-mean}{std}$$

Let us check the *max* and *min* values for the normalized image.

```python
lobster_tensor_sym.min().item(),lobster_tensor_sym.max().item()
```

So, the normalizer in this case worked as *min-max* scaler. Note that the *mean* and *std* are not normalized, indeed we have:

```python
#compute mean and std of the lobster_tensor_norm across three channels
# that is, dropping dim 1 and 2, leaving channels only
lobster_tensor_sym.mean(dim=[1,2]),lobster_tensor_sym.std(dim=[1,2])
```

As we observed before, having mean = $0$ and std = $1$ for features also helps with gradient descent (See Notebook 1) and this can also be achieved using <code>Normalize</code>, as illustrated below.

### Utilizing <code>TransformedTensorDataset</code>

Previously, we used <code>TensorDataset</code>, the simplest possible way to match features with labels. Now, we want to apply transforms to data items and therefore we will build a customized dataset that can apply transformations. To do this, we just need to extend the <code>Dataset</code> class and implement three functions: <code>\_\_init\_\_, \_\_len\_\_</code> and <code>\_\_getitem\_\_</code>. The <code>\_\_getitem\_\_</code> function retrieves a sample from the dataset at the given index <code>idx</code>, applies the transforms, if applicable, and returns the tensor image and corresponding label in a tuple. Note that if no transform is provided the dataset reduces to what we used before, the original <code>TensorDataset</code>.

```python
class TransformedTensorDataset(Dataset):
    def __init__(self, x, y, transform=None):
        self.x = x
        self.y = y
        self.transform = transform
        
    def __getitem__(self, idx):
        x = self.x[idx]

        # apply transforms if provided
        if self.transform:
            x = self.transform(x)
        
        return x, self.y[idx]
        
    def __len__(self):
        return len(self.x)
```

```python
transform = Compose([RandomHorizontalFlip(p=1.0),
                    Normalize(mean=[.5], std=[.5])])
```

### Balanced vs imbalanced dataset

If one label is more common than the others, the dataset is **imbalanced**. As we discussed before (see Notebook 3), in imbalanced datasets, a model can achieve high accuracy by simply predicting mostly the majority class instances, completely ignoring minority classes. Such a model would be biased towards majority classes and perform poorly on unseen examples.
Ideally, each mini-batch should contain multiple minority class examples (preferably preserving class proportions). Batches that don't contain sufficient minority classes will train very poorly. It is usually a good idea to first train on the original dataset and see how the model performs. This gives you a baseline for future comparisons against other experiments you run. Then you can try to alleviate the problem by, for example,  applying **weighted random samplers** to create mini-batches with a sufficient number of minority class items.

Let us check our images dataset's class proportions.

As you can see, we have approximately 3 times more labels 0 than 1 or 2. This may lead to a model that is biased towards the class with label 0 if we do not create suitable mini-batches in training.

### Complete weighted sampler

Let us put it together as a handy function.

```python
%%writefile handy_code/weighted_sampler.py

def weighted_sampler(y, device):
    classes, counts = y.unique(return_counts=True)
    weights = 1.0 / counts.float()
    sample_weights = weights[y.squeeze().long()]
    generator = torch.Generator(device=device)
    sampler = WeightedRandomSampler(
        weights=sample_weights,
        num_samples=len(sample_weights),
        generator=generator,
        replacement=True
    )
    return sampler
```

## Convolutions

We <b>flattened</b> images, treating each pixel value as a feature. Note that by doing this, we ignore the spatial structure of the image, that is, the original location of a pixel in the image. The only thing that counted (considered) was the pixel's value. As you may expect, this was a substantial <b>loss of information</b> contained in the image, i.e. information encoded in the spatial configuration of pixels. How can we make use of this information? One sensible way is to use convolutions between the image and a so called, <b>kernel</b>, or <b>filter</b>.

### Filter (kernel)

We define a *filter*, also known as *kernel*, which is typically a small square matrix like the one below. We apply it to the portion of the image of the same size as the filter (**convolving the image**), starting from the left top corner. The region in gray, to which the filter is being applied, is called the <b>receptive field</b>.

<img align = "left" src="images/conv1.png" width="50%"> 

```python
single = np.array(
    [[[[5, 0, 8, 7, 8, 1],
       [1, 9, 5, 0, 7, 7],
       [6, 0, 2, 4, 6, 6],
       [9, 7, 6, 6, 8, 4],
       [8, 3, 8, 5, 1, 3],
       [7, 2, 7, 0, 1, 0]]]]
)
single.shape
```

```python
identity = np.array(
    [[[[0, 0, 0],
       [0, 1, 0],
       [0, 0, 0]]]]
)
identity.shape
```

**How does it work?**

We multiply each region value by the corresponding filter value (**element-wise** multiplication) and take the sum of these products as the resulting pixel value on the output (the filtered image).

<img align = "left" src="images/conv2.png" width="50%"> 

```python
# in Python
region = single[:, :, 0:3, 0:3]
filtered_region = region * identity
total = filtered_region.sum()
total
```

The outcome:

<img align = "left" src="images/conv3.png" width="50%"> 

Next step: move to the right and re-apply the filter to the new region (in gray).

<img align = "left" src="images/stride1.png" width="50%"> 

```python
new_region = single[:, :, 0:3, (0+1):(3+1)]
```

<img align = "left" src="images/conv4.png" width="50%"> 

```python
new_filtered_region = new_region * identity
new_total = new_filtered_region.sum()
new_total
```

The outcome:

<img align = "left" src="images/conv5.png" width="50%"> 

Next move is forbidden because it would have region partially outside the image and the calculation would fail.

<img align = "left" src="images/conv6.png" width="50%"> 

We continue by going back to the left side and moving down one step. Again, we move to the right as many times as possible (no portion of the receptive field is allowed to be placed outside the image) and continue from the left side moving down by one step. Once all the valid regions are covered, we have the output image. Observe that the output image is smaller than the original one!

<img align = "left" src="images/conv7.png" width="50%"> 

**Calculating the size of the output image**

Our filter moves one step at a time. We say that we use a <b>stride</b> of one. Therefore,
if the input image has $(h,w)$ shape and the filter has $(h_f,w_f)$ shape, the shape of the output image can be found using the formula:

$$
\large
(h, w) * (h_f, w_f) = (h - (h_f - 1), w - (w_f - 1)) = (h - h_f + 1, w - w_f + 1)  
$$

Note that the larger the stride, the smaller the resulting image. Here we assumed stride=1. The general formula will follow once we discuss the concept of *padding*.

### Padding - the way to go outside the image

By adding columns and rows of zeros around the image we make room for the filter to be placed partially outside the original image. This way the calculations can be completed and the output image has the same size as the original or bigger (depends on how much padding you add), if required.

<img align = "left" src="images/padding1.png" width="50%"> 

Depending on the application, different kinds of padding may prove helpful.

**Constant padding**

```python
# convert input image to tensor
image = torch.as_tensor(single).float()
# create padding 
const_padder = nn.ConstantPad2d(padding=2, value=0)
# apply padding
const_padder(image)
```

**Other padding examples**

<img align = "left" src="./images/paddings.png" width="50%"> 

In **replication padding**, pixel values mirror the real pixel values (the ones in dark orange color).

```python
replication_padder = nn.ReplicationPad2d(padding=1)
replication_padder(image)
```

In **reflection padding**, the outer rows and columns are used as axes for reflection (including corners). The values used in the reflection are in dark orange.

```python
reflection_padder = nn.ReflectionPad2d(padding=1)
reflection_padder(image)
```

### 2D Convolution Module 

PyTorch implements convolutions as a **module**  (see [PyTorch convolutions](https://pytorch.org/docs/stable/generated/torch.nn.Conv2d.html)), named <code>Conv2d</code>.

The convolution module takes several arguments, where the positional ones (to be always provided) are:

- <code>in_channels</code>: number of channels of the input image
- <code>out_channels</code>: number of channels produced by the convolution
- <code>kernel_size</code>: size of the convolving kernel (filter)

It also takes a number of (optional) keyword/named arguments like <code>stride</code>, <code>padding</code>, and <code>dilation</code> (spacing beween kernel elements).

### Pooling operation 

Visit: (https://pytorch.org/docs/stable/generated/torch.nn.MaxPool2d.html)

Pooling replaces the output of the network at a certain region, i.e. a portion of an image like a $3\times 3$ square, with some value calculated over that region, such as average, max, weighted average etc. Below you can find an illustration of **max pooling**, where the image is split into $2\times 2$ rectangular blocks, the maximum output within each block is found (in red) and put together forming a $3\times 3$ ouput.

**Max pooling**

Produces a maximal activation value over the receptive field. It is usually connected with the stride of the same size, therefore the stride default value for pooling is the pooling kernel size.

**What is pooling useful for?**

Pooling layers are often used between convolutional layers. They result in:
- reduced computational complexity (smaller images result in less computations)
- reduced memory usage
- reduced number of parameters
  
In consequence, **pooling reduces overfitting**. Additionally, it makes the model invariant to small image shifts.

Let I = input size (square tensor), K = kernel size (square tensor) , S = stride value, and P = padding value. Then the output size is:

### Adding regularization - denying certain features

<div class ="class alert alert-info">
Regularization is any modification we make to a model learning process aimed at the reduction of the model generalization error but not its training error.
</div>    

**Dropout** 

The idea behind using *dropout* is similar to the randomization of features in **random forests** for performing the splits, where each split uses only a subset of features. The dropout meachanism in neural networks allows dropping some features, thus forcing the model to achieve the target in a different way, making it potentially more robust (performing better on unseen datasets). 

We illustrate the technique below by incorporating the *dropout* technique in the above poorly performing model.
Note that by setting $p=0.5$, the average number of dropped elements (neurons (inputs) or filters, if used with the layers) will be $50\%$. 


```python
torch.manual_seed(44)

model_cnn3 = nn.Sequential(nn.Dropout(p=0.5))
n_channels = 1
model_cnn3.add_module('conv1', nn.Conv2d(in_channels=1, out_channels=4, padding=1, kernel_size=3))
model_cnn3.add_module('relu1', nn.ReLU())
model_cnn3.add_module('maxp1', nn.MaxPool2d(kernel_size=2))

model_cnn3.add_module('conv2', nn.Conv2d(in_channels=4, out_channels = 6, padding=1, kernel_size=3))
model_cnn3.add_module('relu2', nn.ReLU())
model_cnn3.add_module('flatten', nn.Flatten())
# dropout layers may also be utilized
#model_cnn3.add_module('dropout', nn.Dropout(p=0.1))

# Classification (add fully connected layer)
model_cnn3.add_module('fc1', nn.Linear(in_features=6*5*5, out_features=10))
model_cnn3.add_module('relu3', nn.ReLU())

# Output Layer
model_cnn3.add_module('fc2', nn.Linear(in_features=10, out_features=3))
```

```python
lr = 0.01
multi_loss_fn = nn.CrossEntropyLoss(reduction='mean')
optimizer_cnn3 = optim.SGD(model_cnn3.parameters(), lr=lr)
mb_cnn3 = ModelBuilder(model_cnn3, multi_loss_fn, optimizer_cnn3,print_loss_freq=5)
mb_cnn3.set_loaders(train_loader, val_loader)
```

```python
mb_cnn3.train(50)
mb_cnn3.plot_losses()
```

```python
model_cnn3.eval()
# get the predictions
test_pred = torch.as_tensor(mb_cnn3.predict(x_test))

# convert predictions to probabilities
test_probab = nn.Softmax(dim=1)(test_pred)

# get predicted class labels
y_test_pred= torch.argmax(test_probab,dim=1)
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
ConfusionMatrixDisplay.from_predictions(y_test.detach().cpu(),y_test_pred,ax=ax)
plt.grid(False)
```

```python
test_correct = (y_test_pred == y_test.cpu()).float()
test_accuracy = test_correct.mean()
test_accuracy
```

## Regularization of neural networks

You are now aware that choosing the optimal size of a network, fully connected or CNN, is a challenge for a developer. There is no clear pathway telling us how many layers or how wide the layers should be to achieve the best performance within reasonable time. Small networks have a low capacity and are likely to **underfit** since they may not be able learn the complex structure of your dataset. On the other hand, large networks may result in **overfitting**, where the network will memorize the training data and perform poorly on the held-out test dataset. When dealing with real-world machine learning problems, we do not know  how deep or how wide the network should be a *priori*. **Regularization techniques** help improve a neural network’s generalization ability by reducing overfitting. 

**Common regularization techniques** 

- **Early stopping** (discussed in Notebook 3, Sec. 3.7); stop training when the validation loss starts increasing.
- **Dropout** (using <code>nn.Droput()</code> layer in training as in the example above; dropped units are selected randomly in each forward-pass of training (but NOT in prediction)). This way the network is forced to learn more general and robust patterns of the data.
- **Data Augmentation** - increase the size of the training set by creating additional new data points. For example, you could add rotated digits to MNIST or impose random noise on images.
- **Adaptive Learning Rate** and **Learning Rate Schedulers** - see Textbook, Chapter 6 for details. 

## What if it does not work?

There is no general rule, but the following hints may help:

- Carefully monitoring training loss; it should decrease on the training set, if it does not your network architecture is wrong.
- Reducing the training and validation set to see whether it works, if not then you know the problem is not the model capacity.
- Trying bare-bones model first (turn off any additional techniques like dropout, augmentation etc.); if it works, gradually add complexity to improve the model. 
- Make sure you train on correct input - for example, it may expect values in $[0,1]$ while you pass $[0,255]$, gray-level image values.
- Check hyperparameters: a too small learning rate may have no impact on outcomes, a too big learning rate may result in big losses.
